## *** ------- Our integration: inspect the unchanged author source ZIP

Our read-only audit checks the saved ZIP's integrity, revision marker, and presence of the relevant released files. It does not extract, modify, or publish benchmark inputs. Run after mounting your own Drive. Failure is a stop signal, not evidence that the model or dataset is invalid.


```python
from pathlib import Path
import hashlib, zipfile

AUTHOR_ZIP = Path('/content/drive/MyDrive/MM_SafetyBench_Full_20261102_PRIVATE/MM-SafetyBench-main.zip')
EXPECTED_REVISION = 'b80eedea3db312c09ded2082813390f68e750ef3'
EXPECTED_SHA256 = '801a891382a06b7c3b29083dd3844b6da6a3e0ef31d3dfdd9086e30b323f322b'
if not AUTHOR_ZIP.is_file():
    print('Source ZIP unavailable: mount Drive or set AUTHOR_ZIP to your verified copy.')
else:
    digest = hashlib.sha256(AUTHOR_ZIP.read_bytes()).hexdigest()
    with zipfile.ZipFile(AUTHOR_ZIP) as archive:
        names = set(archive.namelist())
        needed = ['README.md', 'TinyVersion_ID_List.json', 'evaluation.py', 'creation/1_extract_key_words.py', 'creation/2_img_process.py']
        checks = {item: any(name.endswith('/' + item) for name in names) for item in needed}
        questions = [name for name in names if '/data/processed_questions/' in name and name.endswith('.json')]
        print('ZIP hash matches preserved source:', digest == EXPECTED_SHA256)
        print('Revision marker matches:', archive.comment.decode('ascii', errors='replace') == EXPECTED_REVISION)
        print('ZIP integrity OK:', archive.testzip() is None)
        print('Required source files present:', all(checks.values()))
        print('Processed scenario JSON files:', len(questions))
        print('Source file checks:', checks)
```

## *** ------- Our integration: audit released question and image pair counts

This is our read-only code. It compares author question IDs against the separately downloaded author image ZIP and reports aggregate counts only. The image ZIP path is not yet known to persist in Drive: provide its actual location. This check does not establish that the README TYPO mapping matches the paper or decode every image.


```python
import json, zipfile
from collections import Counter

IMAGE_ZIP = Path('/content/drive/MyDrive/MM_SafetyBench_Full_20261102_PRIVATE/MM-SafetyBench(imgs).zip')
if not AUTHOR_ZIP.is_file() or not IMAGE_ZIP.is_file():
    print('Question or image archive missing. Set IMAGE_ZIP to the verified author archive; no files changed.')
else:
    with zipfile.ZipFile(AUTHOR_ZIP) as source, zipfile.ZipFile(IMAGE_ZIP) as images:
        image_names = images.namelist()
        question_names = [n for n in source.namelist() if '/data/processed_questions/' in n and n.endswith('.json')]
        counts = Counter()
        for name in question_names:
            scenario = Path(name).stem
            for item in json.loads(source.read(name)):
                for variant in ('SD', 'TYPO', 'SD_TYPO'):
                    suffix = f'/{scenario}/{variant}/{item}.jpg'
                    matches = sum(entry.endswith(suffix) for entry in image_names)
                    counts['planned'] += 1
                    counts['exactly_one' if matches == 1 else 'missing' if matches == 0 else 'ambiguous'] += 1
        print('Scenario files:', len(question_names))
        print('Pair counts:', dict(counts))
        print('Image ZIP integrity OK:', images.testzip() is None)
```
